```python
import json

with open('../data/labels.json', 'r') as f:
    d = json.load(f)
```

```python
import re
for item in d['labels']:
    #item['labels']
    matches = re.findall(r'-?\d*\.?\d+', item['path'][:-9])
    item['label'] = [int(matches[0])]
```

```python
with open('../data/labels_good.json', "w") as f:
    json.dump(d, f)
```
